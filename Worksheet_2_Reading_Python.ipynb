## 1. Variables and types (about 5 minutes)

A **variable** is a name that stores a value. In Python `=` means "store this value under this name", not "is equal to".

The main types of value are text in quotes (a *string*, `str`), whole numbers (*integers*, `int`), numbers with a decimal point (*floats*, `float`) and `True`/`False` (*booleans*, `bool`).

```python
drug_name = "Aspirin"
dose_mg = 325.0
tablets = 2
total_mg = dose_mg * tablets
print(drug_name, total_mg)
```

Why does the total print as `650.0` and not `650`?

```python
print("5" + "3")
print(5 + 3)
print("5" * 3)
```

Quotes matter: `"5"` is text and `5` is a number. Numbers stored as text are a common source of silent mistakes in code that reads data from files.

```python
print(type(dose_mg), type(tablets), type(drug_name))
```

## 2. Lists and dictionaries (about 6 minutes)

A **list** keeps items in order. Items are numbered from **0**, not 1, and negative numbers count from the end. A **dictionary** stores `key: value` pairs, and you look a value up by its key.

```python
drugs = ["Aspirin", "Ibuprofen", "Paracetamol", "Metformin"]
print(drugs[0])
print(drugs[-1])
print(len(drugs))
print(drugs[1:3])
```

```python
drug_info = {"name": "Metformin", "dose_mg": 500, "class": "Biguanide"}
print(drug_info["dose_mg"])

drug_info["generic"] = True
print(drug_info)
```

## 3. Decisions with if / elif / else

Code under an `if` runs only when its condition is true. Python checks the conditions from the top and stops at the first one that is true. The indentation (the spaces at the start of a line) matters.

```python
size = 120

if size <= 50:
    print("Nanoparticle is too small")
elif size <= 200:
    print("Nanoparticle is just right")
else:
    print("Nanoparticle is too big")
```

## 4. Loops

A `for` loop repeats the indented code once for each item. `range(1, 4)` starts at 1 and stops **before** 4.

```python
for day in range(1, 4):
    print("Day", day)
```

```python
doses_mg = [250, 250, 500, 500]
total = 0

for dose in doses_mg:
    total = total + dose
    print("Running total:", total)

print("Final:", total)
```

```python
storage_temps = [20, 22, 26, 30, 24, 19]
max_temp = 25

for temp in storage_temps:
    if temp <= max_temp:
        print(f"At {temp} C: Stable")
    else:
        print(f"At {temp} C: Unstable")
```

## 5. Functions

A **function** is a named recipe. `def` starts it, the names in brackets are its **inputs**, and `return` gives back its **output**. Once defined, you can use it as many times as you like.

```python
def calculate_bmi(weight_kg, height_m):
    return weight_kg / height_m ** 2

print(calculate_bmi(70, 1.75))
print(calculate_bmi(70, 175))
```

The second answer is nonsense, but Python did not complain. **Python does not know about units.** A function will happily run with the wrong units and give a wrong answer with no warning. Always ask what units a function expects.

## 6. Libraries

Python on its own has only basic tools. A **library** adds more. `import numpy as np` loads the numpy maths library and lets you refer to it as `np`. Later you will meet `pandas` (tables of data) and `seaborn` (plots).

```python
import numpy as np

numbers = [2, 4, 1, 8, 15]
print(np.sum(numbers))
print(np.mean(numbers))
print(np.sqrt(16))
print(np.abs(-3.5))
```

## 7. Reading code you did not write

This function is written the way an LLM might write it. Do not run it yet. Read it and answer the four questions underneath.

```python
def percent_difference(a_values, b_values):
    """Percentage difference between the average of two dissolution profiles."""
    mean_a = sum(a_values) / len(a_values)
    mean_b = sum(b_values) / len(b_values)
    return abs(mean_a - mean_b) / ((mean_a + mean_b) / 2) * 100
```

1. What does the function take as input?
2. What does it return, and in what units?
3. For `a_values = [20, 40]` and `b_values = [10, 30]`
4. What do you think happens if you give it an empty list?


Now run the next two cells and compare them with your answers.

```python
formulation_A = [20, 40, 60, 75, 85, 95]
formulation_B = [10, 30, 55, 70, 80, 90]

print(percent_difference([20, 40], [10, 30]))
print(percent_difference(formulation_A, formulation_B))
```

## The reading checklist

For any piece of code, ask:

1. **Inputs.** What goes in, and in what units?
2. **Outputs.** What comes out, and in what units?
3. **Edges.** What happens at zero, at a boundary value, or with empty input?
4. **Unknown tools.** Is there a function or library I do not recognise? What does it do?
5. **One case by hand.** Can I work out one answer myself and compare?
